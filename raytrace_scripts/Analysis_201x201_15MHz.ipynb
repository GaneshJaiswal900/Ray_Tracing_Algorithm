```python
# ===================================================
# Electron density model used in raytrace module
# for analytical calculation
# ===================================================

# --------------------------------------------------
# PARAMS OBJECT
# --------------------------------------------------
class Params(object):
    pass

prm          = Params()
prm.r_chromo = 1.02
prm.r_corona = 1.05

# --------------------------------------------------
# ELECTRON DENSITY MODEL
# --------------------------------------------------
def density_only(prm, x, y, z):
    r2 = x*x + y*y + z*z
    r  = math.sqrt(r2)

    global r_corm16, r_corm6, r_corm2d5

    R0 = 6.955e5
    g1 = 3.09e8
    g2 = 1.58e8
    g3 = 2.51e6

    r_corm16  = prm.r_corona**(-16)
    r_corm6   = prm.r_corona**(-6)
    r_corm2d5 = prm.r_corona**(-2.5)

    A = np.zeros((4, 4))
    A[0] = [1., prm.r_chromo, prm.r_chromo**2,
            prm.r_chromo**3]
    A[1] = [1., prm.r_corona, prm.r_corona**2,
            prm.r_corona**3]
    A[2] = [0., 1., 2.*prm.r_chromo,
            3.*prm.r_chromo**2]
    A[3] = [0., 1., 2.*prm.r_corona,
            3.*prm.r_corona**2]
    A = np.linalg.inv(A)

    rhsa    = np.zeros(4)
    rhsa[0] = 5.7e11 * math.exp(
        -7.7e-4*(R0*(prm.r_chromo - 1) - 500))
    a       = np.dot(A, rhsa)

    if r < prm.r_chromo:
        Ne = 5.7e11 * math.exp(
            -7.7e-4*(R0*(r - 1) - 500))
    else:
        rm2d5     = r**(-2.5)
        rm6       = r**(-6)
        rm16      = r**(-16)
        cosTh     = abs(z / r)
        sqrtCosTh = math.sqrt(cosTh)
        t1 = 1.0 - 0.5*cosTh
        t2 = 1.0 - 0.95*cosTh
        t3 = 1.0 - sqrtCosTh

        if r < prm.r_corona:
            saito_rcor = (g1*r_corm16*t1 +
                          g2*r_corm6*t2  +
                          g3*r_corm2d5*t3)
            rhsb    = np.zeros(4)
            rhsb[1] = 1.
            rhsb[2] = (-438.9e6 * R0 *
                       math.exp(-7.7e-4*(
                           R0*(prm.r_chromo-1)-500))
                       / saito_rcor)
            rhsb[3] = ((-16.*g1*(prm.r_corona**(-17))*t1
                        - 6.*g2*(prm.r_corona**(-7))*t2
                        - 2.5*g3*(prm.r_corona**(-3.5))*t3)
                       / saito_rcor)
            b_coeff = np.dot(A, rhsb)
            Ne = ((a[0] + r*(a[1] + r*(a[2] + r*a[3])))
                  + (b_coeff[0] + r*(b_coeff[1] +
                     r*(b_coeff[2] + r*b_coeff[3])))
                  * saito_rcor)
        else:
            Ne = g1*rm16*t1 + g2*rm6*t2 + g3*rm2d5*t3

    return Ne
```

```python
# --------------------------------------------------
# CRITICAL ELECTRON DENSITY FOR 15 MHz
# --------------------------------------------------
me    = 9.10938e-28
e_cgs = 4.80326e-10
pi    = math.pi


mhz_15  = mhz_int(freq)
omega   = 2.0 * pi * freq
Ne_cr   = (me * omega**2) / (4.0 * pi * e_cgs**2)

print("Frequency : {} MHz".format(mhz_15))
print("Ne_cr     : {:.6e} cm^-3".format(Ne_cr))
```

```python
# ===================================================
# Critical surface for 15 MHz at different 
# Y offset plane from central axis
# ===================================================

# --------------------------------------------------
# PHYSICAL Y,Z COORDINATES
# --------------------------------------------------
y_arr = np.linspace(rect[0], rect[2], nx)
z_arr = np.linspace(rect[1], rect[3], ny)
```

```python
# ==================================================
# Y=0 plane
# x search: from outer sphere inward to 0
# For each Z value, vary x inward to find x_crit
# R_crit = sqrt(x_crit^2 + 0^2 + Z^2)
# ==================================================

x_start  = 20.0
x_end    = 0.0
n_steps  = 1000   # fine step size for accuracy

x_arr = np.linspace(x_start, x_end, n_steps)
Y_fixed    = 0
z_vals     = z_arr        # 201 Z values
R_crit_z   = np.full(len(z_vals), np.nan)
x_crit_arr  = np.full(len(z_vals), np.nan)


for j in range(len(z_vals)):
    Zj       = z_arr[j]

    Ne_prev = density_only(prm, x_arr[0], Y_fixed, Zj)

    for k in range(1, len(x_arr)):
        x  = x_arr[k]
        Ne = density_only(prm, x, Y_fixed, Zj)

        if Ne >= Ne_cr:
            # interpolate between x_arr[k-1] and x_arr[k]
            x_prev = x_arr[k-1]
            frac   = (Ne_cr - Ne_prev) / (Ne - Ne_prev + 1e-30)
            x_crit = x_prev + frac * (x - x_prev)

            # critical radius
            R_crit_z[j] = math.sqrt(
                x_crit**2 + Y_fixed**2 + Zj**2)
            x_crit_arr[j] = x_crit
            break

        Ne_prev = Ne

    print("  j={:3d}  Z={:7.4f} Rs  "
          "R_crit={}".format(
              j, Zj,
              "{:.8f} Rs".format(R_crit_z[j])
              if not np.isnan(R_crit_z[j])
              else "NOT FOUND"))

valid = ~np.isnan(R_crit_z)
print("\nFound R_crit for {}/{} rays".format(valid.sum(), z_vals))
print("R_crit min = {:.8f} Rs".format(np.nanmin(R_crit_z)))
print("R_crit max = {:.8f} Rs".format(np.nanmax(R_crit_z)))
print("R_crit std = {:.6e} Rs".format(np.nanstd(R_crit_z)))
```

```python
# --------------------------------------------------
# PLOT: dZ combined twin-axis, single annotation per curve
# at Zi closest to 0.15 / 0.16 R_sun
# --------------------------------------------------

fig, ax_bot = plt.subplots(figsize=(15, 8))
ax_top = ax_bot.twiny()
 
colors     = ['red',   'green',  'blue']
labels     = [r'$Y = 0$ plane',
              r'$Y = 0.5$ $R_{\odot}$ plane',
              r'$Y = 1.0$ $R_{\odot}$ plane']
min_colors = ['darkred', 'darkgreen', 'navy']
 
# Target Zi values — one per dataset (edit as needed)
target_zi  = [0.15, 0.16, 0.15]
 
datasets = [
    (z_initial,   refl_Z,   OUT_DIR_1),
    (z_initial_2, refl_Z_2, OUT_DIR_2),
    (z_initial_3, refl_Z_3, OUT_DIR_3),
]
 
all_minima = []
 
# ── compute, print, and plot each dataset ─────────────────────────────────────
for i, (z_init, refl_Z_i, _) in enumerate(datasets):
    valid    = (~np.isnan(z_init)) & (~np.isnan(refl_Z_i))
    zi_plot  = z_init[valid]
    dZ       = np.abs(refl_Z_i[valid] - zi_plot)
    sort_idx = np.argsort(zi_plot)
    zi_s     = zi_plot[sort_idx]
    dZ_s     = dZ[sort_idx]
 
    # find the single point closest to target_zi[i]
    closest_idx = int(np.argmin(np.abs(zi_s - target_zi[i])))
    xp  = float(zi_s[closest_idx])
    yp  = float(dZ_s[closest_idx])
    lat = float(np.degrees(np.arcsin(np.clip(xp, -1.0, 1.0))))
 
    all_minima.append((xp, yp, lat, i, zi_s, dZ_s))
 
    # print
    print("-- Annotation point {0} ----------------------".format(labels[i]))
    print("   Zi = {0:.2f} R_sun,  dZ = {1:.2f},  lat = {2:.2f} deg".format(
        xp, yp, lat))
 
    # curve
    ax_bot.plot(zi_s, dZ_s,
                linestyle='-', marker='o', markersize=3,
                linewidth=1.8, color=colors[i],
                label=r'$\Delta z$: {0}'.format(labels[i]), zorder=3)
 
    # single scatter marker
    ax_bot.scatter([xp], [yp], s=90, zorder=5,
                   color=min_colors[i], edgecolors='white', linewidths=0.9,
                   label='Point {0}'.format(labels[i]))

 
# ── axis decoration ───────────────────────────────────────────────────────────
ax_bot.set_xlabel(r'$Z_{i}$  ($R_{\odot}$)', fontsize=20)
ax_bot.set_ylabel(r'$\Delta z$ = |$Z_{r}$ - $Z_{i}$| ($R_{\odot}$)', fontsize=20)
ax_bot.grid(True, linestyle='--', alpha=0.4)
ax_bot.tick_params(axis='both', labelsize=16)
ax_bot.legend(fontsize=12, loc='upper left',
              bbox_to_anchor=(1.01, 1.0), borderaxespad=0, ncol=1)
 
# upper twin axis
zi_lo, zi_hi = ax_bot.get_xlim()
ax_top.set_xlim(zi_lo, zi_hi)
lat_ticks_deg = np.array([-90, -60, -45, -30, -20, -10,
                            0, 10, 20, 30, 45, 60, 90], dtype=float)
zi_for_ticks  = np.sin(np.radians(lat_ticks_deg))
in_range = (zi_for_ticks >= max(zi_lo, -1.0)) & \
           (zi_for_ticks <= min(zi_hi,  1.0))
ax_top.set_xticks(zi_for_ticks[in_range])
ax_top.set_xticklabels(
    ['{0:.0f}'.format(l) for l in lat_ticks_deg[in_range]], fontsize=16)
ax_top.set_xlabel(r'Latitude  $\phi$ (degree)', fontsize=20, labelpad=8)
 
# ── occupancy grid for 3 label placements ────────────────────────────────────
plt.subplots_adjust(top=0.82, bottom=0.12, left=0.10, right=0.78)
fig.canvas.draw()
 
x0, x1   = ax_bot.get_xlim()
y0, y1   = ax_bot.get_ylim()
y1_safe  = y0 + (y1 - y0) * 0.95
 
NX, NY   = 60, 50
occupied = np.zeros((NY, NX), dtype=bool)
 
# mark curve cells occupied (buffer +-2)
for (xp_c, yp_c, lat_c, ic, zi_s_c, dZ_s_c) in all_minima:
    for px, py in zip(zi_s_c, dZ_s_c):
        cx = int((px - x0) / (x1 - x0) * NX)
        cy = int((py - y0) / (y1_safe - y0) * NY)
        for dcx in range(-2, 3):
            for dcy in range(-2, 3):
                nx2 = cx + dcx
                ny2 = cy + dcy
                if 0 <= nx2 < NX and 0 <= ny2 < NY:
                    occupied[ny2, nx2] = True
 
# mark below-zero region occupied
for cy in range(NY):
    cy_val = y0 + (cy + 0.5) / NY * (y1_safe - y0)
    if cy_val <= 0.0:
        occupied[cy, :] = True
 
lbl_ncx = max(2, int(NX * 0.18))
lbl_ncy = max(2, int(NY * 0.20))
 
def block_free(cx0, cy0):
    if cx0 < 0 or cy0 < 0:
        return False
    if cx0 + lbl_ncx > NX or cy0 + lbl_ncy > NY:
        return False
    return not np.any(occupied[cy0:cy0 + lbl_ncy, cx0:cx0 + lbl_ncx])
 
def mark_block(cx0, cy0):
    x0b = max(0, cx0 - 1)
    y0b = max(0, cy0 - 1)
    x1b = min(NX, cx0 + lbl_ncx + 1)
    y1b = min(NY, cy0 + lbl_ncy + 1)
    occupied[y0b:y1b, x0b:x1b] = True
 
def find_nearest_free_block(xp, yp):
    cx_ideal = int((xp - x0) / (x1 - x0) * NX) - lbl_ncx // 2
    cy_ideal = int((yp - y0) / (y1_safe - y0) * NY) - lbl_ncy // 2
    for radius in range(max(NX, NY)):
        for dcx in range(-radius, radius + 1):
            for dcy in range(-radius, radius + 1):
                if abs(dcx) != radius and abs(dcy) != radius:
                    continue
                cx0c = cx_ideal + dcx
                cy0c = cy_ideal + dcy
                if block_free(cx0c, cy0c):
                    return (cx0c, cy0c)
    return (0, NY - lbl_ncy)
 
# ── annotate exactly 3 points ────────────────────────────────────────────────
for (xp, yp, latp, i, zi_s, dZ_s) in all_minima:
 
    lbl = ('$Z_i$={0:.2f} $R_\\odot$\n'
           '$\\Delta z$={1:.2f}\n'
           '$\\phi$={2:.2f} deg').format(xp, yp, latp)
 
    cx0, cy0 = find_nearest_free_block(xp, yp)
    mark_block(cx0, cy0)
 
    xl = x0 + (cx0 + lbl_ncx * 0.5) / NX * (x1 - x0)
    yl = y0 + (cy0 + lbl_ncy * 0.5) / NY * (y1_safe - y0)
 
    ax_bot.annotate(
        lbl,
        xy=(xp, yp),
        xytext=(xl, yl),
        textcoords='data',
        fontsize=12,
        color=min_colors[i],
        ha='center', va='center',
        bbox=dict(boxstyle='round,pad=0.30',
                  fc='white', ec=min_colors[i],
                  alpha=0.95, lw=0.9),
        arrowprops=dict(arrowstyle='->',
                        color=min_colors[i],
                        lw=1.0,
                        shrinkA=4, shrinkB=4,
                        connectionstyle='arc3,rad=0.15'),
        zorder=15
    )
 
# ── save ──────────────────────────────────────────────────────────────────────
plt.savefig(os.path.join(OUT_DIR,
            "delta_Z_vs_Zinit_twin_15mhz_combined.png"),
            dpi=150, bbox_inches='tight')
plt.show()
plt.close()
print("Saved combined dZ plot.")
```

```python
# ======================================================
# BENDING ANGLE: initial direction vs FINAL direction
# (after the ray has fully turned and is heading away)
# Y=0.0
# ======================================================

# --------------------------------------------------
# Initial direction vector for each ray
# --------------------------------------------------
v_initial = np.full((ny, 3), np.nan)

for j in range(ny):
    p0 = traj[j, 0, :]
    p1 = traj[j, 1, :]

    delta = p1 - p0
    norm  = np.linalg.norm(delta)
    if norm > 1e-12:
        v_initial[j] = delta / norm

# --------------------------------------------------
# Final direction vector for each ray
# --------------------------------------------------
v_final = np.full((ny, 3), np.nan)
idx_final = np.full(ny, -1, dtype=int)

for j in range(ny):
    r_trace = np.sqrt(traj[j, :, 0]**2 +
                      traj[j, :, 1]**2 +
                      traj[j, :, 2]**2)
    nonzero      = r_trace > 0.01
    valid_steps  = np.where(nonzero)[0]

    if len(valid_steps) < 2:
        continue

    # Last valid step index
    idx_last     = valid_steps[-1]
    idx_final[j] = idx_last

    # Use last two valid steps for final direction
    # (one-sided backward difference at the end)
    i1 = idx_last
    i0 = valid_steps[-2]

    delta = traj[j, i1, :] - traj[j, i0, :]
    norm  = np.linalg.norm(delta)
    if norm > 1e-12:
        v_final[j] = delta / norm


# --------------------------------------------------
# Bending angle
# --------------------------------------------------
angle_deg = np.full(ny, np.nan)

for j in range(ny):
    vi = v_initial[j]
    vf = v_final[j]

    if not (np.any(np.isnan(vi)) or np.any(np.isnan(vf))):
        cos_theta    = np.clip(np.dot(vi, vf), -1.0, 1.0)
        angle_deg[j] = np.degrees(np.arccos(cos_theta))

valid_angle = ~np.isnan(angle_deg)

# Find ray closest to 90 degrees
diff_from_90 = np.abs(angle_deg - 90.0)
idx_90       = np.nanargmin(diff_from_90)
print("\n  Closest to 90 deg:")
print("    j={}  angle={:.4f} deg".format(idx_90, angle_deg[idx_90]))
print("    Z_initial = {:.4f} Rs".format(z_initial[idx_90]))
print("    Z_refl    = {:.4f} Rs".format(refl_Z[idx_90]))


# ==================================================
# PLOT 1: Bending angle vs Z_initial
# ==================================================
fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(z_initial[valid_angle], angle_deg[valid_angle],
        linestyle='-', marker='o', markersize=4,
        linewidth=1.8, color='purple',
        label=r'$\delta$')

ax.axhline(90.0, color='red', linestyle='--',
           linewidth=1.8, label=r'$\delta$=90 deg')

ax.axhline(0.0, color='gray', linestyle=':',
           linewidth=1.2, alpha=0.6, label=r'$\delta$=0 deg')

ax.plot(z_initial[j_zero], angle_deg[j_zero],
        marker='*', markersize=14, color='darkviolet',
        zorder=10,
        label='Z=0 ray: {:.2f} deg'.format(angle_deg[j_zero]))

ax.plot(z_initial[idx_90], angle_deg[idx_90],
        marker='D', markersize=9, color='red',
        zorder=10,
        label= r'$\delta$ close to 90 deg: $Z_i$={:.4f}'.format(
              z_initial[idx_90]))

ax.set_xlabel(r'$Z_i$ ($R_{\odot}$)', fontsize=12)
ax.set_ylabel(r'$\delta$(degrees)', fontsize=12)
ax.set_title(r'$\delta$ vs $Z_i$',
             fontsize=11)
ax.legend(fontsize=9)
ax.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR_1,
            'bending_angle_init_vs_final_Zinit_15mhz.png'), dpi=150)
plt.show()
plt.close()
print("Saved bending angle vs Z_initial.")

# ==================================================
# PLOT 2: Bending angle vs Z_reflection
# ==================================================
fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(refl_Z[valid_angle], angle_deg[valid_angle],
        linestyle='-', marker='o', markersize=4,
        linewidth=1.8, color='darkorange',
        label=r'$\delta$')

ax.axhline(90.0, color='red', linestyle='--',
           linewidth=1.8, label=r'$\delta$=90 deg')

ax.axhline(0.0, color='gray', linestyle=':',
           linewidth=1.2, alpha=0.6, label=r'$\delta$=0 deg')

ax.plot(refl_Z[j_zero], angle_deg[j_zero],
        marker='*', markersize=14, color='darkorange',
        zorder=10,
        label='Z=0 ray: {:.2f} deg'.format(angle_deg[j_zero]))

ax.plot(refl_Z[idx_90], angle_deg[idx_90],
        marker='D', markersize=9, color='red',
        zorder=10,
        label=r'$\delta$ close to 90 deg: $Z_r$={:.4f}'.format(
              refl_Z[idx_90]))

ax.set_xlabel(r'$Z_r$ ($R_{\odot}$)', fontsize=12)
ax.set_ylabel(r'$\delta$(degrees)', fontsize=12)
ax.set_title(r'$\delta$ vs $Z_r$',
             fontsize=11)
ax.legend(fontsize=9)
ax.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR_1,
            'bending_angle_init_vs_final_Zrefl_15mhz.png'), dpi=150)
plt.show()
plt.close()
print("Saved bending angle vs Z_reflection.")



# --------------------------------------------------
# Basic counts
# --------------------------------------------------
total_valid   = valid_angle.sum()

above_90 = valid_angle & (angle_deg >  90.0)
below_90 = valid_angle & (angle_deg <  90.0)
exact_90 = valid_angle & (angle_deg == 90.0)

n_above  = above_90.sum()
n_below  = below_90.sum()
n_exact  = exact_90.sum()

pct_above = 100.0 * n_above / total_valid
pct_below = 100.0 * n_below / total_valid
pct_exact = 100.0 * n_exact / total_valid

print("\n--- Ray bending statistics (15 MHz) ---")
print("  Total valid rays      : {}".format(total_valid))
print("  Angle > 90 deg        : {}  ({:.2f} %)".format(n_above, pct_above))
print("  Angle < 90 deg        : {}  ({:.2f} %)".format(n_below, pct_below))
print("  Angle = 90 deg exactly: {}  ({:.2f} %)".format(n_exact, pct_exact))


# ==================================================
# SAVE TABLE IMAGE
# ==================================================

fig, ax = plt.subplots(figsize=(8, 5))
ax.axis('off')

main_data = [
    ['Total valid rays',      str(total_valid),   '100.00 %'],
    [r'$\delta$ > 90 deg',        str(n_above),        '{:.2f} %'.format(pct_above)],
    [r'$\delta$ < 90 deg',        str(n_below),        '{:.2f} %'.format(pct_below)],
    [r'$\delta$=90 deg exactly',str(n_exact),        '{:.2f} %'.format(pct_exact)],
]
col_labels = ['Category', 'Ray Count', 'Percentage']

tbl = ax.table(
    cellText  = main_data,
    colLabels = col_labels,
    cellLoc   = 'center',
    loc       = 'center'
)
tbl.auto_set_font_size(False)
tbl.set_fontsize(11)
tbl.scale(1.4, 2.2)

# Header row style
for col in range(len(col_labels)):
    tbl[0, col].set_facecolor('#2c3e50')
    tbl[0, col].set_text_props(color='white', fontweight='bold')

# Row styles
row_styles = {
    1: ('#fadbd8', '#922b21'),   # above 90 -- light red
    2: ('#d6eaf8', '#1a5276'),   # below 90 -- light blue
    3: ('#d5f5e3', '#1e8449'),   # exact 90 -- light green
    4: ('#fdfefe', '#2c3e50'),   # total    -- white (won't appear but safe)
}
# Total row (row index 1 in our data is "Total valid rays" -- restyle it)
for col in range(len(col_labels)):
    # Total valid rays row
    tbl[1, col].set_facecolor('#eaecee')
    tbl[1, col].set_text_props(color='#2c3e50', fontweight='bold')
    # Above 90
    tbl[2, col].set_facecolor('#fadbd8')
    tbl[2, col].set_text_props(color='#922b21')
    # Below 90
    tbl[3, col].set_facecolor('#d6eaf8')
    tbl[3, col].set_text_props(color='#1a5276')
    # Exact 90
    tbl[4, col].set_facecolor('#d5f5e3')
    tbl[4, col].set_text_props(color='#1e8449')

ax.set_title(r'$\delta$ Statistics Y=0.0',
             fontsize=13, fontweight='bold', pad=16)

plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR_1,
            'bending_stats_table_15mhz.png'),
            dpi=150, bbox_inches='tight')
plt.show()
plt.close()
print("Saved bending statistics table image.")
```

```python
# ==================================================
# Plot: Delta Z vs X
# For two rays:
# Ray 1: (i,j)=(100,133)  Z_init ~ 0.5 Rs
# Ray 2: (i,j)=(100,167)  Z_init ~ 1.0 Rs
# ==================================================

i1, j1 = 100, 133
i2, j2 = 100, 167

idx1   = i1 * ny + j1
idx2   = i2 * ny + j2

traj1  = traj_all[idx1, :, :]   # (Nsteps, 3)
traj2  = traj_all[idx2, :, :]   # (Nsteps, 3)

print("Ray 1: (i,j)=({},{})  "
      "Launch X={:.4f}  Z={:.4f}".format(
      i1, j1, traj1[0,0], traj1[0,2]))
print("Ray 2: (i,j)=({},{})  "
      "Launch X={:.4f}  Z={:.4f}".format(
      i2, j2, traj2[0,0], traj2[0,2]))

z_arr_1 = traj1[:, 2]
z_arr_2 = traj2[:, 2]

x_arr = traj1[:, 0]

delta_z = z_arr_2 - z_arr_1


# --------------------------------------------------
# Find reflection point for both ray
# i.e. point of closest approach to Sun center
# --------------------------------------------------
def get_reflection_point(traj):
    x_t = traj[:, 0]
    y_t = traj[:, 1]
    z_t = traj[:, 2]
    r_trace = np.sqrt(x_t**2 + y_t**2 + z_t**2)
    nonzero = r_trace > 0.01
    if nonzero.sum() > 0:
        full_indices = np.where(nonzero)[0]
        idx_min  = np.argmin(r_trace[nonzero])
        idx_full = full_indices[idx_min]
        return x_t[idx_full], z_t[idx_full], r_trace[idx_full]
    return np.nan, np.nan, np.nan

refl_X1, refl_Z1, refl_R1 = get_reflection_point(traj1)
refl_X2, refl_Z2, refl_R2 = get_reflection_point(traj2)

print("\nRay 1 reflection point:  X={:.4f}  Z={:.4f}  R={:.4f}".format(
    refl_X1, refl_Z1, refl_R1))
print("Ray 2 reflection point:  X={:.4f}  Z={:.4f}  R={:.4f}".format(
    refl_X2, refl_Z2, refl_R2))

# --------------------------------------------------
# Plot
# --------------------------------------------------
fig, ax = plt.subplots(figsize=(9, 5))

# Original data
ax.plot(x_arr, delta_z,
        linestyle='-', marker='o', markersize=4,
        linewidth=1.8, color='purple',
        label=r'$\Delta$Z vs X')

ax.plot(refl_X1, refl_Z1,
        linestyle='-', marker='o', markersize=5,
        linewidth=1.8, color='red',
        label=r'Reflection point $R_1$')

ax.plot(refl_X2, refl_Z2,
        linestyle='-', marker='o', markersize=5,
        linewidth=1.8, color='green',
        label=r'Reflection point $R_2$')

ax.set_xlabel(r'X ($R_{\odot}$)', fontsize=12)
ax.set_ylabel(r'$\Delta$Z', fontsize=12)
ax.set_title(r'$\Delta$Z vs X ($Z_1$=0.5$R_{\odot}$ and $Z_2$=1.0$R_{\odot}$)', fontsize=11)
ax.set_xlim(1,4)
ax.legend(fontsize=9)
ax.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR,
            'delta_z_vs_x_15mhz.png'), dpi=150)
plt.show()
plt.close()
print("Saved fit plot")
```

```python
# ==================================================
# Plot: Angle between v and grad(n)/n vs X
# For two rays:
# Ray 1: (i,j)=(100,133)  Z_init ~ 0.5 Rs
# Ray 2: (i,j)=(100,167)  Z_init ~ 1.0 Rs
# ==================================================
def refr_index_n(Ne, Ne_cr):
    n2 = 1.0 - Ne / Ne_cr
    if n2 <= 0.0:
        return 0.0
    return math.sqrt(n2)


def grad_n_over_n_at(prm, x, y, z, Ne_cr, h=1e-4):
    """
    grad(n)/n = -grad(Ne) / (2*(Ne_cr - Ne))
    Eq.4 of paper
    """
    def Ne_at(xx, yy, zz):
        return density_only(prm, xx, yy, zz)

    Ne0   = Ne_at(x, y, z)
    denom = 2.0 * (Ne_cr - Ne0)
    if abs(denom) < 1e-30:
        return np.array([0.0, 0.0, 0.0])

    dNe_dx = (Ne_at(x+h, y,   z) - Ne_at(x, y, z)) / h
    dNe_dy = (Ne_at(x,   y+h, z) - Ne_at(x, y, z)) / h
    dNe_dz = (Ne_at(x,   y,   z+h) - Ne_at(x, y, z)) / h

    grad_Ne = np.array([dNe_dx, dNe_dy, dNe_dz])
    return -grad_Ne / denom


# --------------------------------------------------
# COMPUTE ANGLE BETWEEN v AND grad(n)/n
# AT EVERY STEP ALONG THE RAY
# --------------------------------------------------
def angle_v_gnn_along_ray(traj_ray, prm, Ne_cr):
    r_trace     = np.sqrt(traj_ray[:, 0]**2 +
                          traj_ray[:, 1]**2 +
                          traj_ray[:, 2]**2)
    nonzero     = r_trace > 0.01
    valid_steps = np.where(nonzero)[0]

    if len(valid_steps) < 2:
        return None, None, None

    idx_refl = valid_steps[np.argmin(r_trace[nonzero])]
    x_refl   = traj_ray[idx_refl, 0]

    x_coords = []
    angles   = []

    for k in range(idx_refl):
        pos = traj_ray[k, :]
        xk  = float(pos[0])
        yk  = float(pos[1])
        zk  = float(pos[2])
        rk  = math.sqrt(xk**2 + yk**2 + zk**2)

        if rk < 0.01:
            continue

        # Ray direction at step k
        pos_next = traj_ray[k+1, :]
        delta    = pos_next - pos
        norm     = np.linalg.norm(delta)
        if norm < 1e-15:
            continue
        v_k = delta / norm

        # grad(n)/n at this position
        try:
            gnn = grad_n_over_n_at(prm, xk, yk, zk, Ne_cr)
        except Exception:
            continue

        mag_gnn = np.linalg.norm(gnn)
        if mag_gnn < 1e-30:
            continue

        gnn_hat   = gnn / mag_gnn
        cos_theta = np.clip(np.dot(v_k, gnn_hat), -1.0, 1.0)
        theta_deg = math.degrees(math.acos(cos_theta))

        x_coords.append(xk)
        angles.append(theta_deg)

    return np.array(x_coords), np.array(angles), x_refl


# --------------------------------------------------
# COMPUTE ANGLES ALONG BOTH RAYS
# --------------------------------------------------
print("\nComputing angle(v, grad(n)/n) along Ray 1...")
x1, ang1, xrefl1 = angle_v_gnn_along_ray(traj1, prm, Ne_cr)
print("  Steps: {0}  X_refl={1:.4f} Rs".format(len(x1), xrefl1))

print("Computing angle(v, grad(n)/n) along Ray 2...")
x2, ang2, xrefl2 = angle_v_gnn_along_ray(traj2, prm, Ne_cr)
print("  Steps: {0}  X_refl={1:.4f} Rs".format(len(x2), xrefl2))


# ==================
# PLOT: Angle vs X  
# ==================
fig, ax = plt.subplots(figsize=(10, 7))

# --- Ray curves ---
ax.plot(x1, ang1,
        linestyle='-', linewidth=2.0, color='blue',
        label=r'Ray 1: $Z_i$={0:.4f} $R_{{\odot}}$  '
              r'(start={1:.1f} deg, end={2:.1f} deg)'.format(
                  traj1[0, 2], ang1[0], ang1[-1]))

ax.plot(x2, ang2,
        linestyle='-', linewidth=2.0, color='red',
        label=r'Ray 2: $Z_i$={0:.4f} $R_{{\odot}}$  '
              r'(start={1:.1f} deg, end={2:.1f} deg)'.format(
                  traj2[0, 2], ang2[0], ang2[-1]))

# --- Reflection points ---
ax.axvline(xrefl1, color='blue', linestyle='--',
           linewidth=1.5, alpha=0.7,
           label=r'Ray 1 reflection  $X$={0:.4f} $R_{{\odot}}$'.format(
               xrefl1))

ax.axvline(xrefl2, color='red', linestyle='--',
           linewidth=1.5, alpha=0.7,
           label=r'Ray 2 reflection  $X$={0:.4f} $R_{{\odot}}$'.format(
               xrefl2))

# --- Analytical X_crit ---
if not np.isnan(x_crit_arr[j1]):
    ax.axvline(x_crit_arr[j1],
               color='blue', linestyle=':', linewidth=1.5, alpha=0.8,
               label=r'Ray 1 $X_c$={0:.4f} $R_{{\odot}}$'.format(
                   x_crit_arr[j1]))

if not np.isnan(x_crit_arr[j2]):
    ax.axvline(x_crit_arr[j2],
               color='red', linestyle=':', linewidth=1.5, alpha=0.8,
               label=r'Ray 2 $X_c$={0:.4f} $R_{{\odot}}$'.format(
                   x_crit_arr[j2]))

# --- Axes labels and limits ---
ax.set_xlabel(r'$X$ ($R_{\odot}$)', fontsize=28)
ax.set_ylabel(r'$\gamma$ (deg)',     fontsize=28)
plt.tick_params(axis='both', labelsize=16)
ax.set_ylim(80, 190)
ax.grid(True, linestyle='--', alpha=0.4)

# --- Legend: outside, centred above the axes ---
ax.legend(
    fontsize=16,
    loc='lower center',
    bbox_to_anchor=(0.5, 1.02),
    ncol=2,
    frameon=True,
    framealpha=0.9,
    edgecolor='gray',
    borderaxespad=0.0
)

plt.tight_layout()
plt.savefig(
    os.path.join(OUT_DIR, 'angle_v_gnn_vs_X_two_rays_15mhz.png'),
    dpi=150,
    bbox_inches='tight'
)
plt.show()
plt.close()
print("Saved Plot 1: angle(v, grad(n)/n) vs X.")
```

```python
# ==================================================
# HELPER: combined difference plot with twin x-axis
# upper x-axis: phi = arcsin(Zi) in degrees
# ==================================================

def plot_diff_combined(datasets, xlabel, ylabel, fname, out_dir):
    colors_dark = {'red': 'darkred', 'blue': 'darkblue', 'green': 'darkgreen'}

    fig, ax_bot = plt.subplots(figsize=(12, 5))  # wider figure to accommodate outside legend

    for ds in datasets:
        z_axis     = ds['z_axis']
        diff_vals  = ds['diff_vals']
        valid_mask = ds['valid_mask']
        j_zero     = ds['j_zero']
        color      = ds['color']
        lbl        = ds['label']

        ax_bot.plot(z_axis[valid_mask], diff_vals[valid_mask],
                    linestyle='-', marker='o', markersize=4,
                    linewidth=1.8, color=color, label=lbl)

        if valid_mask[j_zero]:
            z_zero_val    = z_axis[j_zero]
            diff_zero_val = diff_vals[j_zero]
            dark          = colors_dark.get(color, color)
            ax_bot.plot(z_zero_val, diff_zero_val,
                        marker='*', markersize=14,
                        color=dark, zorder=10,
                        label='Z=0 ray ({0}): {1:.2f} Rs'.format(lbl, diff_zero_val))

    # --------------------------------------------------
    # Bottom x-axis
    # --------------------------------------------------
    ax_bot.set_xlabel(xlabel, fontsize=20)
    ax_bot.set_ylabel(ylabel, fontsize=16)
    ax_bot.grid(True, linestyle='--', alpha=0.4)

    # Legend outside on the right
    ax_bot.legend(
        fontsize=9,
        loc='upper left',
        bbox_to_anchor=(1.02, 1),   # just outside right edge
        borderaxespad=0,
        frameon=True,
        framealpha=0.9,
        edgecolor='gray'
    )

    # --------------------------------------------------
    # Top x-axis: phi = arcsin(Zi) in degrees
    # --------------------------------------------------
    ax_top = ax_bot.twiny()

    x_min, x_max = ax_bot.get_xlim()
    ax_top.set_xlim(x_min, x_max)

    phi_ticks_deg = np.array([-90, -60, -45, -30, -15,
                               0, 15, 30, 45, 60, 90])
    zi_ticks = np.sin(np.radians(phi_ticks_deg))

    in_range = (zi_ticks >= x_min) & (zi_ticks <= x_max)
    zi_ticks_plot  = zi_ticks[in_range]
    phi_ticks_plot = phi_ticks_deg[in_range]

    ax_top.set_xticks(zi_ticks_plot)
    ax_top.set_xticklabels(
        ['{0:.0f}'.format(p) for p in phi_ticks_plot],
        fontsize=10
    )
    ax_top.set_xlabel(r'Latitude $\phi$ (degrees)', fontsize=14)

    # bbox_inches='tight' ensures the legend is not clipped when saving
    plt.tight_layout()
    plt.savefig(os.path.join(out_dir, fname), dpi=150, bbox_inches='tight')
    plt.show()
    plt.close()
    print("Saved: {}".format(fname))


    # ==================================================
# COMBINED PLOT CALL
# ==================================================
plot_diff_combined(
    datasets=[
        {
            'z_axis'    : z_initial,
            'diff_vals' : diff_Y,
            'valid_mask': valid_both,
            'j_zero'    : j_zero,
            'color'     : 'green',
            'label'     : r'$\Delta Y$ ($Y = 0.0$ plane)',
        },
        {
            'z_axis'    : z_initial_2,
            'diff_vals' : diff_Y_2,
            'valid_mask': valid_both,
            'j_zero'    : j_zero,
            'color'     : 'blue',
            'label'     : r'$\Delta Y$ ($Y = 0.5$ $R_{\odot}$ plane)',
        },
        {
            'z_axis'    : z_initial_3,
            'diff_vals' : diff_Y_3,
            'valid_mask': valid_both,
            'j_zero'    : j_zero,
            'color'     : 'red',
            'label'     : r'$\Delta Y$ ($Y = 1.0$ $R_{\odot}$ plane)',
        },
    ],
    xlabel  = r'$Z_{i}$ ($R_{\odot}$)',
    ylabel  = r'$\Delta Y$ = $Y_{r}$ - $Y_{i}$ ($R_{\odot}$)',
    fname   = 'dY_vs_Zinit_15mhz_combined.png',
    out_dir = OUT_DIR,
)


# ==================================================
# COMBINED PLOT CALL
# ==================================================
plot_diff_combined(
    datasets=[
        {
            'z_axis'    : z_initial,
            'diff_vals' : refl_H,
            'valid_mask': valid_both,
            'j_zero'    : j_zero,
            'color'     : 'green',
            'label'     : r'$H_r$ ($Y = 0.0$ plane)',
        },
        {
            'z_axis'    : z_initial_2,
            'diff_vals' : refl_H_2,
            'valid_mask': valid_both_2,
            'j_zero'    : j_zero,
            'color'     : 'blue',
            'label'     : r'$H_r$ ($Y = 0.5$ $R_{\odot}$ plane)',
        },
        {
            'z_axis'    : z_initial_3,
            'diff_vals' : refl_H_3,
            'valid_mask': valid_both_3,
            'j_zero'    : j_zero,
            'color'     : 'red',
            'label'     : r'$H_r$ ($Y = 1.0$ $R_{\odot}$ plane)',
        },
    ],
    xlabel  = r'$Z_{i}$ ($R_{\odot}$)',
    ylabel  = r'$H_r$ = $R_{r}$ - 1 ($R_{\odot}$)',
    fname   = 'H_vs_Zinit_15mhz_combined.png',
    out_dir = OUT_DIR,
)
    
```
