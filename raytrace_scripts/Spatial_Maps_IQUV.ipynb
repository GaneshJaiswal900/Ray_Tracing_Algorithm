```python
from __future__ import print_function, division
import os
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

SIM_DIR  = "/home/hp/raytrace_work/raytrace_results/Simulation_Data_201x201"
IQUV_DIR = "/home/hp/raytrace_work/raytrace_results/Spatial_Maps_IQUV_201x201"

if not os.path.exists(IQUV_DIR):
    os.makedirs(IQUV_DIR)

params    = np.load(os.path.join(SIM_DIR, "sim_params.npy"),
                    allow_pickle=True).item()
freq_list = np.array(params['freq_list'])
n_freqs   = len(freq_list)

def mhz_int(f):
    return int(round(f / 1.e6))

extent        = [-2, 2, -2, 2]
stokes_titles = [u"Stokes I", u"Stokes Q",
                 u"Stokes U", u"Stokes V"]
stokes_keys   = [u"I", u"Q", u"U", u"V"]

for freq_idx, freq_hz in enumerate(freq_list):

    mhz  = mhz_int(freq_hz)
    fl   = "{} MHz".format(mhz)
    fdir = os.path.join(IQUV_DIR, "{}mhz".format(mhz))
    if not os.path.exists(fdir):
        os.makedirs(fdir)

    path = os.path.join(
        SIM_DIR, "tbriquv_{}mhz.npy".format(mhz))
    if not os.path.exists(path):
        print("Missing: {}".format(path))
        continue

    print("[{:02d}/15] {} MHz".format(freq_idx+1, mhz))
    tbriquv = np.load(path)   # (nx, ny, 4)

    # --------------------------------------------------
    # PLOT 1: 2x2 combined IQUV map
    # --------------------------------------------------
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))
    axes = axes.flatten()
    for k in range(4):
        im = axes[k].imshow(
            tbriquv[:, :, k],
            extent=extent,
            origin=u"lower")
        axes[k].set_title(
            u"{} - {}".format(stokes_titles[k], fl))
        axes[k].set_xlabel(u"Solar Radii")
        axes[k].set_ylabel(u"Solar Radii")
        axes[k].set_xticks([-2, -1, 0, 1, 2])
        axes[k].set_yticks([-2, -1, 0, 1, 2])
        plt.colorbar(im, ax=axes[k],
                     fraction=0.046, pad=0.04,
                     label=u"Tb (K)")
    plt.suptitle(
        u"Stokes IQUV Spatial Maps - {}".format(fl),
        fontsize=14)
    plt.tight_layout()
    plt.savefig(os.path.join(
        fdir, "IQUV_2x2_{}mhz.png".format(mhz)),
        dpi=300, bbox_inches=u"tight")
    plt.close()

    # --------------------------------------------------
    # PLOT 2: Individual Stokes maps
    # --------------------------------------------------
    for k, key in enumerate(stokes_keys):
        fig, ax = plt.subplots(figsize=(6, 5))
        im = ax.imshow(
            tbriquv[:, :, k],
            extent=extent,
            origin=u"lower")
        ax.set_title(
            u"Stokes {} - {}".format(key, fl))
        ax.set_xlabel(u"Solar Radii")
        ax.set_ylabel(u"Solar Radii")
        ax.set_xticks([-2, -1, 0, 1, 2])
        ax.set_yticks([-2, -1, 0, 1, 2])
        plt.colorbar(im, ax=ax,
                     fraction=0.046, pad=0.04,
                     label=u"Tb (K)")
        plt.tight_layout()
        plt.savefig(os.path.join(
            fdir,
            "stokes{}_{}.png".format(
                key, "{}mhz".format(mhz))),
            dpi=300, bbox_inches=u"tight")
        plt.close()

    # --------------------------------------------------
    # SAVE CSV
    # --------------------------------------------------
    for k, key in enumerate(stokes_keys):
        csv_path = os.path.join(
            fdir,
            "stokes{}_{}.csv".format(
                key, "{}mhz".format(mhz)))
        data = tbriquv[:, :, k]
        with open(csv_path, "wb") as f:
            for r in range(data.shape[0]):
                f.write(",".join(
                    ["{:.6e}".format(data[r, c])
                     for c in range(data.shape[1])])
                    + "\n")

    print("  Saved to: {}".format(fdir))
    del tbriquv

print("\nAll done. Results: {}".format(IQUV_DIR))
```
